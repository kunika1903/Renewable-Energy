## Derived Business Metrics

```python
# Calculate Profit

df['Profit'] = (
    df['Revenue']
    - df['Operating_Cost']
    - df['Maintenance_Cost']
)

df[['Revenue', 'Operating_Cost', 'Maintenance_Cost', 'Profit']].head()
```

```python
# Calculate Profit Margin

df['Profit_Margin_Percent'] = (
    df['Profit'] / df['Revenue']
) * 100

df[['Revenue', 'Profit', 'Profit_Margin_Percent']].head()
```

```python
# Energy generated per MW of installed capacity

df['Generation_per_MW'] = (
    df['Energy_Generated_MWh'] /
    df['Installed_Capacity_MW']
)

df[['Installed_Capacity_MW',
   'Energy_Generated_MWh',
   'Generation_per_MW']].head()
```

```python
# Revenue generated per MW of installed capacity

df['Revenue_per_MW'] = (
    df['Revenue'] /
    df['Installed_Capacity_MW']
)

df[['Installed_Capacity_MW',
   'Revenue',
   'Revenue_per_MW']].head()
```

```python
# Calculate cost efficiency

df['Cost_Efficiency'] = (
    df['Revenue'] /
    (df['Operating_Cost'] + df['Maintenance_Cost'])
)

df[['Revenue',
   'Operating_Cost',
   'Maintenance_Cost',
   'Cost_Efficiency']].head()
```

```python
# Overall business metrics

total_energy = df['Energy_Generated_MWh'].sum()
total_revenue = df['Revenue'].sum()
total_operating_cost = df['Operating_Cost'].sum()
total_maintenance_cost = df['Maintenance_Cost'].sum()
total_profit = df['Profit'].sum()

print("Total Energy Generated:", total_energy)
print("Total Revenue:", total_revenue)
print("Total Operating Cost:", total_operating_cost)
print("Total Maintenance Cost:", total_maintenance_cost)
print("Total Profit:", total_profit)
```

```python
# Average performance metrics

print("Average Efficiency:",
      df['Efficiency_Percent'].mean())

print("Average Availability:",
      df['Availability_Percent'].mean())

print("Average Capacity Utilization:",
      df['Capacity_Utilization'].mean())

print("Average Plant Performance Score:",
      df['Plant_Performance_Score'].mean())

print("Average Downtime:",
      df['Downtime_Hours'].mean())
```

```python
energy_type_metrics = df.groupby('Energy_Type').agg({
    'Energy_Generated_MWh': 'sum',
    'Revenue': 'sum',
    'Operating_Cost': 'sum',
    'Maintenance_Cost': 'sum',
    'Profit': 'sum',
    'Efficiency_Percent': 'mean',
    'Plant_Performance_Score': 'mean'
}).round(2)

energy_type_metrics
```

## Data Visualization

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

### Energy Generation by Energy Type

```python
energy_generation = df.groupby('Energy_Type')['Energy_Generated_MWh'].sum()

plt.figure(figsize=(8,5))

plt.bar(
    energy_generation.index,
    energy_generation.values
)

plt.title('Total Energy Generated by Energy Type')
plt.xlabel('Energy Type')
plt.ylabel('Energy Generated (MWh)')

plt.show()
```

#### Insight: Compare Solar, Wind and Hydro based on total energy generation.

### Which Region Generates the Most Revenue?

```python
region_analysis = df.groupby('Region').agg({
    'Energy_Generated_MWh': 'sum',
    'Revenue': 'sum',
    'Profit': 'sum',
    'Operating_Cost': 'sum',
    'Maintenance_Cost': 'sum',
    'Efficiency_Percent': 'mean',
    'Plant_Performance_Score': 'mean',
    'Downtime_Hours': 'mean'
}).round(2)

region_analysis
```

#### Find the highest-revenue region:

```python
highest_revenue_region = (
    region_analysis['Revenue']
    .idxmax()
)

highest_revenue = (
    region_analysis['Revenue']
    .max()
)

print("Highest Revenue Region:", highest_revenue_region)
print("Revenue:", highest_revenue)
```

#### Find the energy type with the highest downtime:

```python
highest_downtime_type = (
    operational_analysis['Downtime_Hours']
    .idxmax()
)

print("Highest Downtime Energy Type:", highest_downtime_type)
```
